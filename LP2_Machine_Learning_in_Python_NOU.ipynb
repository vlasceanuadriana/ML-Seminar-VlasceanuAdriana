## 6. Vizualizarea datelor — tipuri de grafice

Vom explora, pe rând, cele mai utile tipuri de grafice pentru un proiect de machine learning: distribuții pentru o singură variabilă, relații între două sau mai multe variabile și diagrame pentru date categoriale sau relaționale.

### 6.1 `countplot` — distribuția unei variabile categoriale

Afișează numărul de observații pentru fiecare categorie.

````python
sns.countplot(x='gender', data=galton_df)
````

````python
# adăugăm o a doua variabilă, prin culoare (hue)
sns.countplot(x='gender', hue='childNum', data=galton_df)
````

````python
# schimbăm axele: variabila categorială pe verticală
sns.countplot(y='childNum', hue='gender', data=galton_df)
````

### 6.2 `barplot` — compararea unei valori numerice pe grupuri

Afișează estimările punctuale (de regulă, media) și intervalul de încredere (implicit, 95%) ca bare dreptunghiulare.

````python
sns.barplot(x="gender", y="childHeight", data=galton_df)
````

````python
sns.barplot(x="childNum", y="childHeight", data=galton_df)
````

````python
sns.barplot(x="childNum", y="childHeight", hue="gender", data=galton_df)
````

````python
sns.barplot(x="childHeight", y="gender", data=galton_df)
````

### 6.3 `displot` — distribuția unei variabile numerice

`displot()` este o funcție de nivel înalt din Seaborn care creează histograme sau diagrame de densitate.

- **x** — coloana din setul de date folosită pe axa X
- **data** — sursa datelor
- **kde** — `True`/`False`: adaugă sau nu o linie de densitate (*Kernel Density Estimate*) peste histogramă
- **rug** — `True`/`False`: adaugă un *rug plot* ("covoraș") — mici liniuțe pe axa X care indică poziția fiecărui punct individual, utile pentru a vedea concentrarea valorilor brute

````python
sns.displot(x="childHeight", data=galton_df, kde=False)
````

````python
sns.displot(x="childHeight", data=galton_df, kde=True, rug=True)
````

### 6.4 Relații bivariate — `lmplot` și `relplot`

**`lmplot`** ("*linear model plot*") desenează un scatter plot și aplică o linie de regresie liniară (cu interval de încredere) — arată relația dintre două variabile numerice și cât de bine poate fi aproximată printr-o dreaptă.

````python
sns.lmplot(x="mother", y="childHeight", data=galton_df)
````

**Argumentul `hue`** colorează punctele în funcție de o categorie — fiecare valoare a variabilei respective apare cu o culoare diferită.

````python
# reprezentarea grafică a trei variabile simultan (mother, childHeight, gender)
sns.lmplot(x="mother", y="childHeight", hue="gender", data=galton_df)
````

**`relplot`** este o funcție mai flexibilă pentru relații între mai multe variabile, inclusiv continue.

````python
sns.relplot(x="mother", y="childHeight", hue="gender", data=galton_df)
````

**Argumentul `style`**, la fel ca `hue`, diferențiază categorii — dar prin formă sau tip de linie, nu prin culoare.

````python
sns.relplot(x="mother", y="childHeight", hue="childNum", style="gender", data=galton_df)
````

**Argumentul `kind`** controlează tipul de grafic generat, în funcție de funcție:

| Funcție | Ce controlează `kind` | Exemple de valori |
| --- | --- | --- |
| `sns.displot()` | tip de distribuție | `"hist"`, `"kde"`, `"ecdf"` |
| `sns.catplot()` | tip de grafic categorial | `"box"`, `"violin"`, `"bar"`, `"strip"`, `"swarm"` |
| `sns.relplot()` | tip de relație | `"scatter"`, `"line"` |

````python
# variabilă continuă, reprezentată ca linie
sns.relplot(x="childNum", y="childHeight", kind="line", data=galton_df)
````

````python
# aceleași variabile, reprezentate ca puncte (scatter)
sns.relplot(x="childNum", y="childHeight", kind="scatter", data=galton_df)
````

### 6.5 `jointplot` — relație combinată cu distribuții marginale

Combină un grafic principal (scatter sau altă relație între X și Y) cu histograme/densități pentru fiecare variabilă, pe margini — vezi simultan distribuția fiecărei variabile și corelația dintre ele.

````python
sns.jointplot(x="mother", y="childHeight", data=galton_df)
````

**Argumentul `kind`:**

| `kind` | Ce face |
| --- | --- |
| `"scatter"` | scatter plot standard — punctele individuale |
| `"reg"` | scatter plot + dreapta de regresie |
| `"kde"` | curbă de densitate bidimensională (hartă termică) |
| `"hex"` | hexbin plot — culoare mai deschisă/închisă = densitate mai mică/mare |

````python
# adaugă pe grafic funcția de densitate și dreapta de regresie
sns.jointplot(x="mother", y="childHeight", data=galton_df, kind="reg")
````

````python
# diagramă hexbin — folosește o culoare mai deschisă sau mai închisă pentru densitatea datelor
sns.jointplot(x="mother", y="childHeight", data=galton_df, kind="hex")
````

````python
sns.jointplot(x="mother", y="childHeight", data=galton_df, kind="kde")
````

### 6.6 `pairplot` — relații perechi ale întregului set de date

Reprezintă toate distribuțiile bivariate posibile, pe perechi de variabile numerice din setul de date.

````python
sns.pairplot(galton_df)
````

````python
sns.pairplot(galton_df, hue='gender')
````

````python
# pe diagonală putem alege tipul de grafic afișat (implicit era 'kde')
sns.pairplot(galton_df, hue='gender', diag_kind='hist')
````

### 6.7 `catplot` — distribuții categoriale

Reprezintă grafic variabile categoriale, unde fiecare valoare corespunde unui punct.

**Opțiunea `kind`:**

| `kind` | Ce face |
| --- | --- |
| `"strip"` | plasează toate punctele individuale deasupra fiecărei categorii |
| `"swarm"` | ca `strip`, dar punctele nu se suprapun (mai ușor de citit pentru seturi mici) |
| `"box"` | box plot clasic (mediană, IQR, outlieri) |
| `"violin"` | combină box plot cu o curbă de densitate |

````python
sns.catplot(x='childNum', y='childHeight', data=galton_df, kind='strip')
````

````python
sns.catplot(x='childNum', y='childHeight', data=galton_df, kind='swarm')
````

````python
sns.catplot(x='childNum', y='childHeight', data=galton_df, kind='box')
````

````python
sns.catplot(x='childNum', y='gender', data=galton_df, kind='strip')
````

````python
sns.catplot(x='childNum', y='childHeight', data=galton_df, hue='gender', kind='swarm')
````

````python
sns.catplot(x='childNum', y='childHeight', data=galton_df, hue='gender', kind='box')
````

````python
sns.catplot(x='childNum', y='childHeight', kind='violin', data=galton_df)
````

````python
sns.catplot(x='gender', y='childHeight', kind='violin', data=galton_df)
````

### 6.8 Diagrame ierarhice — bar și pie

Cele mai populare diagrame pentru afișarea structurii unei variabile: *pie chart*, *donut chart*, *bar chart*.

````python
galton_df.groupby('gender')['childHeight'].mean().plot.bar()
````

````python
galton_df['gender'].value_counts().plot.pie(
    autopct='%1.1f%%', startangle=90, ylabel='', title='Distribuția copiilor pe gen'
)
````

### 6.9 Diagrame relaționale — `heatmap`

Am văzut deja relații între variabile în diagramele de distribuție (`sns.lmplot()`). O `heatmap` (hartă termică) este utilă pentru a vizualiza fie o tabelă pivotată, fie direct matricea de corelații dintre variabile.

````python
df_pivot = galton3_df.pivot_table(index="mother", columns="father", values="childHeight")
sns.heatmap(df_pivot, annot=True)
````

**Heatmap pentru matricea de corelații**

````python
# calculăm matricea de corelație (doar pentru variabilele numerice)
corr = galton3_df.corr(numeric_only=True)

# heatmap pentru matricea de corelații
sns.heatmap(corr, annot=True)
````
